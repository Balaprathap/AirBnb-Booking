## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.
df['reviews_per_month'] = df['reviews_per_month'].fillna(df['reviews_per_month'].median())
df['last_review'] = df['last_review'].fillna('no review')
df['name'] = df['name'].fillna('No Name')
df['host_name'] = df['host_name'].fillna('No Host')
````

````python
# Remove duplicate rows
df = df.drop_duplicates()
````

````python
# Convert 'last_review' to datetime type
df['last_review'] = pd.to_datetime(df['last_review'], errors='coerce')

# Convert categorical columns to 'category' type for efficiency
df['room_type'] = df['room_type'].astype('category')
df['neighbourhood_group'] = df['neighbourhood_group'].astype('category')
````

````python
# Remove rows where price is greater than 1000 (example of handling outliers)
df = df[df['price'] <= 1000]
````

````python
# Standardize column names (convert to lowercase and replace spaces with underscores)
df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_')
````

````python
# Display the cleaned dataset
print(df.info())
print(df.head())
````

### What all manipulations have you done and insights you found?

I undertook an array of data wrangling procedures to tidy and ready the data set for exploratory data analysis. To begin with, I treated the missing values, noting that the reviews_per_month column would be filled with the median value, the last_review column would be filled with a note ('No Reviews'), and the missing values in the name and host_name columns would be filled with 'No Name' and 'No Host' respectively. I then eliminated duplicate rows, so the data would not have any doubles, and validated the data types were appropriate. I changed the last_review column to a datetime type, and the categorical columns (room_type and neighbourhood_group) were converted to 'category' types. I also removed outliers from the data set, noting I would remove any rows with a price greater than 1000, assuming they were unwanted values. I standardized the column names for convenience, and I created a couple of new features for exploration; for example, I created a price_per_person by dividing the price by the number of reviews. Unwanted features, like id and host_id, were removed from the data set, and I reset the index for convention. These manipulations offered an overview into missing data values, price distributions, review activity metrics, and geographical location. The data set now has proper data types, and it is ready for exploratory analysis of its price trends, availability trends, and/or geo-spatial analysis. These results are important for understanding the influence property has on user interactions in the data set.

### Based on your chart experiments, define three hypothetical statements from the dataset. In the next three questions, perform hypothesis testing to obtain final conclusion about the statements through your code and statistical testing.

There is a noticeable difference between “Entire home/apt” and “Private room” in terms of the mean price of an Airbnb listing. Room type is associated with counting the number of reviews. Room type is associated with neighbourhood group, as well.

### Hypothetical Statement - 1

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

There is no significant difference in the average price between “Entire home/apt” and “Private room” listings.
There is a significant difference in the average price between the two room types.


#### 2. Perform an appropriate statistical test.

````python
from scipy.stats import ttest_ind
# Perform Statistical Test to obtain P-Value
filtered_df = df[df['room_type'].isin(['Entire home/apt', 'Private room'])]

# Optional: Remove extreme price outliers (e.g., above $1000)
filtered_df = filtered_df[filtered_df['price'] < 1000]

# Step 5: Extract price data
entire_prices = filtered_df[filtered_df['room_type'] == 'Entire home/apt']['price']
private_prices = filtered_df[filtered_df['room_type'] == 'Private room']['price']

# Step 6: Perform the two-sample t-test
t_stat, p_value = ttest_ind(entire_prices, private_prices, equal_var=False)

# Step 7: Display results
print(f"T-statistic: {t_stat}")
print(f"P-value: {p_value}")

# Step 8: Interpretation
alpha = 0.05
if p_value < alpha:
    print("Conclusion: Statistically significant difference in prices (Reject Null Hypothesis)")
else:
    print(" Conclusion: No significant difference in prices (Fail to Reject Null Hypothesis)")

# Step 9: Visualize price distributions
plt.figure(figsize=(8,5))
sns.boxplot(x='room_type', y='price', data=filtered_df)
plt.title('Price Comparison Between Room Types')
plt.xlabel('Room Type')
plt.ylabel('Price ($)')
plt.grid(True)
plt.show()
````

##### Which statistical test have you done to obtain P-Value?

We used a **Welch’s two-sample t-test** to compare the mean prices between "Entire home/apt" and "Private room" listings.

##### Why did you choose the specific statistical test?

We chose **Welch’s t-test** because it compares the means of two independent groups **without assuming equal variances**, making it more reliable for real-world data like Airbnb listings, which often have unequal sample sizes and variability.

### Hypothetical Statement - 2

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

The mean number of reviews is the same across all room types.
At least one room type has a significantly different mean number of reviews.

#### 2. Perform an appropriate statistical test.

````python
# Perform Statistical Test to obtain P-Value
from scipy.stats import f_oneway
# Step 4: Check available room types
print("Room Types:", df['room_type'].unique())

# Step 5: Extract number_of_reviews for each room type
entire_reviews = df[df['room_type'] == 'Entire home/apt']['number_of_reviews']
private_reviews = df[df['room_type'] == 'Private room']['number_of_reviews']
shared_reviews = df[df['room_type'] == 'Shared room']['number_of_reviews']

# Step 6: Perform one-way ANOVA test
f_stat, p_value = f_oneway(entire_reviews, private_reviews, shared_reviews)

# Step 7: Display result
print(f"F-statistic: {f_stat}")
print(f"P-value: {p_value}")

# Step 8: Interpretation
alpha = 0.05
if p_value < alpha:
    print(" Conclusion: Significant difference in number of reviews between room types (Reject H₀).")
else:
    print(" Conclusion: No significant difference in number of reviews between room types (Fail to reject H₀).")

# Step 9: Boxplot visualization
plt.figure(figsize=(8,5))
sns.boxplot(x='room_type', y='number_of_reviews', data=df)
plt.title('Number of Reviews by Room Type')
plt.xlabel('Room Type')
plt.ylabel('Number of Reviews')
plt.grid(True)
plt.show()
````
